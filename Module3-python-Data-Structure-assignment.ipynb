Q11. What advantages do dictionaries offer over lists for certain tasks?

Ans:Dictionaries and lists are both fundamental data structures in Python, but they are designed for different purposes. Dictionaries offer several advantages over lists for certain tasks, especially when dealing with key-value pairs and the need for efficient lookups. Below are the key advantages dictionaries have over lists:

1. Faster Lookups
Dictionaries provide average O(1) time complexity for lookups, thanks to their underlying hash table implementation.
In contrast, lists require O(n) time to search for an element, as you may need to iterate through the entire list.

2. Key-Based Access
Dictionaries allow direct access to values using keys that are meaningful (e.g., strings, numbers, or tuples), making the code more readable and intuitive.
Lists rely on numeric indices, which are less descriptive and can make code harder to understand.

3. No Need to Maintain Order
Dictionaries do not require us to maintain a specific order for lookups, which can simplify tasks when the order of elements doesn't matter.
With lists, one must often iterate through the entire structure or sort it to achieve the desired order.

4. Avoid Duplicate Keys
Dictionaries automatically ensure that keys are unique, overwriting the value if the same key is added again. This makes them useful for scenarios where uniqueness is required.
Lists allow duplicate entries, so we must manually ensure uniqueness when needed.

5. Improved Semantics
Dictionaries inherently map keys to values, making them ideal for use cases involving associations, mappings, or structured data (e.g., JSON-like structures).
Lists are better suited for sequences or collections of similar items, where a mapping isn't necessary.

6. Efficient Membership Testing
The in keyword for dictionaries checks keys in O(1) average time.
For lists, the same operation requires O(n) time, as each element must be checked.

7. Customizability with Complex Keys
Dictionaries allow keys to be complex objects like tuples, enabling more advanced mappings.
Lists are limited to simple positional indexing.

```python
'''Q14. How does the 'in' keyword work differently for lists and dictionaries?

Ans:The in keyword is used to check for the presence of an element, but it works differently for lists and dictionaries
because of the different data structures and their underlying implementations.

1. Using in with a List
When you use the in keyword with a list, Python checks whether the specified value exists in the list.
The operation involves iterating through the entire list until the element is found (or until the end of the list is reached).

Time Complexity: O(n), where n is the number of elements in the list.

Search Behavior: It checks for value membership.

2.When you use in with a dictionary, the behavior is different. By default, the in keyword checks for the
presence of the key, not the value. This is because dictionaries store data as key-value pairs.

Time Complexity: O(1) on average, because dictionaries are implemented using hash tables that allow for fast key lookups.

Search Behavior: It checks for key membership.  The above is illustrated by a code as follows:'''

my_dict = {"apple": 3, "banana": 5, "cherry": 2}

# Checking if a key is in the dictionary
print("banana" in my_dict)  # True (key "banana" is in the dictionary)
print("orange" in my_dict)  # False (key "orange" is not in the dictionary)

my_list = [1, 2, 3, 4, 5]

# Checking if a value is in the list
print(3 in my_list)  # True (value 3 is in the list)
print(6 in my_list)  # False (value 6 is not in the list)

```

```python
'''Q16. what is a nested dictionary and give an example of its use case?

Ans:A nested dictionary is a dictionary in which one or more of its values are themselves dictionaries.
Essentially, a dictionary can contain other dictionaries as values, allowing for the representation of
more complex, hierarchical data structures.

Use Case for Nested Dictionaries:
Nested dictionaries are useful when you need to represent data that has multiple levels of organization,
such as configurations, records, or nested data from APIs.

Example Use Case: Storing Student Information
Let's say we are building a student management system, where you want to store information about multiple students.
Each student has various attributes like name, age, and grades for different subjects. This hierarchical data can be
represented using a nested dictionary.

The above can be illustrated by a code as below:'''

# Nested dictionary representing student information
students = {
    "student1": {
        "name": "Alice",
        "age": 20,
        "grades": {
            "math": 90,
            "science": 85,
            "english": 88
        }
    },
    "student2": {
        "name": "Bob",
        "age": 22,
        "grades": {
            "math": 75,
            "science": 80,
            "english": 78
        }
    },
    "student3": {
        "name": "Charlie",
        "age": 21,
        "grades": {
            "math": 92,
            "science": 89,
            "english": 91
        }
    }
}

# Accessing nested dictionary values
print(students["student1"]["name"])  # Output: Alice
print(students["student2"]["grades"]["math"])  # Output: 75

# Adding a new grade for a student
students["student3"]["grades"]["history"] = 95

# Print updated data
print(students["student3"]["grades"])  # Output: {'math': 92, 'science': 89, 'english': 91, 'history': 95}

```

Q17. Describe the time complexity of accessing elements in a dictionary?

Ans:In Python, dictionaries are implemented using hash tables, which provide efficient key-based access. The time complexity for accessing elements in a dictionary is generally O(1) on average, meaning it takes constant time to look up a value given a key. However, the actual performance can vary depending on several factors, which we'll discuss below.

Average Time Complexity: O(1)

Direct key lookup in a dictionary is generally O(1). This means the time it takes to retrieve a value based on its key does not depend on the size of the dictionary.
This is possible because dictionaries use hash tables, where the keys are hashed to produce a unique hash value. The hash value allows Python to directly access the bucket where the value is stored, without needing to search through all the keys.

Worst-Case Time Complexity: O(n)
In the worst case, the time complexity for accessing a dictionary element could be O(n), where n is the number of items in the dictionary. This can happen if there are hash collisions, meaning multiple keys hash to the same bucket, and Python needs to resolve the collisions (typically by using techniques like chaining or open addressing).

If the hash table is poorly distributed (e.g., many keys collide and end up in the same bucket), it could degrade the performance to linear time. However, this is rare in practice because Python uses an effective hashing mechanism.

Example of a worst-case scenario:
If hash collisions happen, the dictionary might need to search through multiple keys in the same bucket, which results in O(n) time complexity in extreme cases, but such cases are uncommon.

Amortized Time Complexity: O(1)
On average, amortized O(1) time complexity for dictionary operations holds in most practical cases, as Python’s hash table implementation handles collisions and resizing efficiently.

 Dictionary Access is Efficient (O(1) on Average):
Hashing: When we access a dictionary with a key, Python computes the hash of the key, which allows it to quickly locate the corresponding value in constant time.
Dynamic Resizing: Python dictionaries resize their internal storage dynamically as they grow, ensuring that the average load factor (the number of items per bucket) remains low, reducing the chance of collisions and maintaining efficient access.
Collision Handling: Python uses efficient collision resolution techniques like chaining or open addressing to minimize the impact of hash collisions on access time.

Q19. Why are dictionaries considered unordered and how does that affect data retrieval?

Ans: Dictionaries in Python are considered unordered because, until Python 3.7, the order of the items in a dictionary was not guaranteed to be the same as the order in which they were added. Even in Python 3.7 and beyond, dictionaries maintain insertion order (i.e., they keep track of the order in which elements are added), but they are still technically unordered in terms of how the internal data is structured and accessed.


The unordered nature of dictionaries affect data retrieval in the following way:

1.Fast Lookup by Key:

Despite being unordered, dictionaries still provide efficient access to values via keys. The O(1) average time complexity for lookups remains unaffected by the order of elements. This is because dictionaries use hash tables, and the lookup operation only depends on the key's hash value, not the position or order of insertion.

2.No Direct Indexing:

Because dictionaries are unordered (internally implemented as hash tables), we cannot directly index into them using a numeric index like you would with lists.
This means you cannot retrieve items by their "position" in the dictionary.

3.Iteration Over Keys:

While dictionaries maintain the order of insertion in Python 3.7 and above, iteration over keys does not provide a predictable sequence of access like indexing in lists.
Iterating over a dictionary gives the keys in the order they were inserted (starting from Python 3.7), but this behavior is an artifact of the insertion order, not a natural property of the dictionary.

4.Performance Considerations:

Accessing by key remains fast and efficient, regardless of the order.
Ordering operations like sorting keys, or obtaining ordered views of items (e.g., items(), keys(), values()), are supported but require extra processing since the dictionary itself is not inherently ordered by design.
